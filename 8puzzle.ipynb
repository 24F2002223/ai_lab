```python
from collections import deque

INITIAL_STATE = (
    (1, 2, 3),
    (0, 4, 6),
    (7, 5, 8)
)

GOAL_STATE = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)

MOVES = [(-1, 0), (1, 0), (0, -1), (0, 1)]

def find_blank(state):
    for r in range(3):
        for c in range(3):
            if state[r][c] == 0:
                return r, c

def get_neighbors(state):
    r, c = find_blank(state)
    neighbors = []

    for dr, dc in MOVES:
        nr, nc = r + dr, c + dc

        if 0 <= nr < 3 and 0 <= nc < 3:
            new_state = [list(row) for row in state]

            new_state[r][c], new_state[nr][nc] = (
                new_state[nr][nc],
                new_state[r][c]
            )

            neighbors.append(tuple(tuple(row) for row in new_state))

    return neighbors

def print_board(state):
    for row in state:
        print(" ".join(str(cell) if cell != 0 else "_" for cell in row))
    print()

def dfs(start, goal, max_depth=50):
    stack = [(start, [start], 0)]
    visited = {start: 0}

    while stack:
        current, path, depth = stack.pop()

        if current == goal:
            return path

        if depth < max_depth:
            for neighbor in get_neighbors(current):
                if neighbor not in visited or visited[neighbor] > depth + 1:
                    visited[neighbor] = depth + 1
                    stack.append(
                        (neighbor, path + [neighbor], depth + 1)
                    )

    return None

def depth_limited_search(current, goal, depth, path, visited):
    if current == goal:
        return path

    if depth <= 0:
        return None

    visited.add(current)

    for neighbor in get_neighbors(current):
        if neighbor not in visited:
            result = depth_limited_search(
                neighbor,
                goal,
                depth - 1,
                path + [neighbor],
                visited
            )

            if result is not None:
                return result

    visited.remove(current)

    return None

def iterative_deepening_search(start, goal, max_depth=20):
    for depth in range(max_depth + 1):
        visited = set()

        result = depth_limited_search(
            start,
            goal,
            depth,
            [start],
            visited
        )

        if result is not None:
            print(f"Goal found at Depth: {depth}")
            return result

    return None

if __name__ == "__main__":
    print("Initial State:")
    print_board(INITIAL_STATE)

    print("Goal State:")
    print_board(GOAL_STATE)

    print("--- Solving using Iterative Deepening Search (IDS) ---")

    ids_solution = iterative_deepening_search(
        INITIAL_STATE,
        GOAL_STATE
    )

    if ids_solution:
        print(
            f"IDS found optimal solution in "
            f"{len(ids_solution) - 1} steps:"
        )

        for step, state in enumerate(ids_solution):
            print(f"Step {step}:")
            print_board(state)
    else:
        print("IDS failed to find a solution within max depth.")

    print("--- Solving using Depth-First Search (DFS) ---")

    dfs_solution = dfs(
        INITIAL_STATE,
        GOAL_STATE
    )

    if dfs_solution:
        print(
            f"DFS found solution in "
            f"{len(dfs_solution) - 1} steps."
        )
    else:
        print("DFS failed to find a solution.")
```
